```python
from hapi import *
import numpy as np
import matplotlib.pyplot as plt

db_begin("data")

fetch("CO2", 2, 1, 2200, 2400)
fetch("O2", 7, 1, 2200, 2400)
fetch("N2", 22, 1, 2200, 2400)
```

```python
fraction_N2 = 0.70
fraction_O2 = 0.20
fraction_CO2 = 0.10

total_fraction = fraction_N2 + fraction_O2 + fraction_CO2

print("Total fraction =", total_fraction) # to check when i later change the composition
```

```python
total_pressure = 1.0

pressure_N2 = fraction_N2 * total_pressure
pressure_O2 = fraction_O2 * total_pressure
pressure_CO2 = fraction_CO2 * total_pressure

print("N2 pressure =", pressure_N2)
print("O2 pressure =", pressure_O2)
print("CO2 pressure =", pressure_CO2)
```

```python
T = 296

nu_CO2, coef_CO2 = absorptionCoefficient_Voigt(
    SourceTables="CO2",
    OmegaRange=[2200, 2400],
    HITRAN_units=False,
    Environment={
        "T": T,
        "p": pressure_CO2
    }
)

nu_O2, coef_O2 = absorptionCoefficient_Voigt(
    SourceTables="O2",
    OmegaRange=[2200, 2400],
    HITRAN_units=False,
    Environment={
        "T": T,
        "p": pressure_O2
    }
)

nu_N2, coef_N2 = absorptionCoefficient_Voigt(
    SourceTables="N2",
    OmegaRange=[2200, 2400],
    HITRAN_units=False,
    Environment={
        "T": T,
        "p": pressure_N2
    }
)
```
